```python
d = df.drop_duplicates().copy()
d["date"] = pd.to_datetime(d["date"])
d = d.sort_values("date").reset_index(drop=True)

for col in d.select_dtypes(include=np.number).columns:
    groups = d[col].ne(d[col].shift()).cumsum()
    runs = d.groupby(groups).agg(
        value=(col, "first"),
        count=(col, "size"),
        first_date=("date", "min"),
        last_date=("date", "max")
    )
    runs = runs[runs["count"] >= 3].sort_values("count", ascending=False)

    if not runs.empty:
        print(f"\n{col}:")
        print(runs.head(5).to_string())
```

```python
d = df.drop_duplicates().copy()
d["date"] = pd.to_datetime(d["date"])
d = d.sort_values(["date", "record_id"], kind="stable").reset_index(drop=True)

for col in ["fodder_price_rs", "temp_c", "feed_kg", "litres_paid_in_passbook"]:
    groups = d[col].ne(d[col].shift()).cumsum()
    runs = d.groupby(groups).agg(
        value=(col, "first"),
        count=(col, "size"),
        first_date=("date", "min"),
        last_date=("date", "max")
    )
    runs = runs[runs["count"] >= 2].sort_values("count", ascending=False)

    print(f"\n{col}:")
    print(runs.head(5).to_string(index=False))
```

```python
d = df.drop_duplicates().copy()
d["date"] = pd.to_datetime(d["date"])
d = d.sort_values(["village_cluster", "date", "record_id"])

for col in d.select_dtypes(include=np.number).columns:
    d["run"] = d.groupby("village_cluster")[col].transform(
        lambda x: x.ne(x.shift()).cumsum()
    )
    runs = d.groupby(["village_cluster", "run"]).agg(
        value=(col, "first"),
        count=(col, "size"),
        first_date=("date", "min"),
        last_date=("date", "max")
    )
    runs = runs[runs["count"] >= 4].sort_values("count", ascending=False)

    if not runs.empty:
        print(f"\n{col}:")
        print(runs.head(5).to_string())
```

```python
d = df.drop_duplicates().sort_values("record_id").reset_index(drop=True)

for col in d.select_dtypes(include=np.number).columns:
    groups = d[col].ne(d[col].shift()).cumsum()
    runs = d.groupby(groups).agg(
        value=(col, "first"),
        count=(col, "size"),
        first_id=("record_id", "first"),
        last_id=("record_id", "last"),
        first_date=("date", "first"),
        last_date=("date", "last")
    )
    longest = runs.loc[runs["count"].idxmax()]
    print(f"{col}: {longest.to_dict()}")
```

```python
x = df.drop_duplicates().sort_values(["village_cluster", "date", "record_id"])

for col in ["temp_c", "feed_kg", "fodder_price_rs"]:
    runs = []
    for cluster, g in x.groupby("village_cluster"):
        s = g[col].ne(g[col].shift()).cumsum()
        for _, r in g.groupby(s):
            if len(r) >= 3:
                runs.append((cluster, r[col].iloc[0], len(r),
                             r["date"].iloc[0], r["date"].iloc[-1]))
    print(col, runs)
```

```python
x = df.drop_duplicates().sort_values(
    ["village_cluster", "date", "record_id"]
)

for col in x.select_dtypes(include="number").columns:
    found = []
    for cluster, g in x.groupby("village_cluster"):
        groups = g[col].ne(g[col].shift()).cumsum()
        for _, r in g.groupby(groups):
            if len(r) >= 3:
                found.append({
                    "cluster": cluster,
                    "value": r[col].iloc[0],
                    "count": len(r),
                    "first_date": r["date"].iloc[0],
                    "last_date": r["date"].iloc[-1]
                })
    if found:
        print(f"\n{col}:")
        print(pd.DataFrame(found).to_string(index=False))
```
